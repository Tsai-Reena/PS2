```python
# 客製化資料集 (Google Map)
import json

# 載入 sample_data.json 文件
with open('./sample_data.json', 'r', encoding='utf-8') as file:
    sample_data = json.load(file)

# 創建新的字典，以符合 id_to_body.json 格式
data = {}
for index, item in enumerate(sample_data, start=1):
    # 使用自定義的鍵，例如 "/comments/{index}"
    key = f'/comments/{index}'
    value = item.get('評論內容', '')
    data[key] = value

# 檢查結果
print(f'資料集大小：{len(data)}')
print(f'範例鍵值：{list(data.keys())[:3]}')
print(f'範例內容：{list(data.values())[:3]}')
```

```python
# 客製化資料集
from tensorflow.keras.preprocessing.text import Tokenizer

# 設定我們想要的詞彙數量
num_words = 10000

# 初始化並擬合 tokenizer
tokenizer = Tokenizer(num_words=num_words)
tokenizer.fit_on_texts(data.values())

# 將文本轉換為編碼
encoded_data = tokenizer.texts_to_sequences(data.values())
word_index = tokenizer.word_index

# 確認結果
print(f'字典大小：{len(word_index)}')
print(f'範例編碼：{encoded_data[:3]}')
```

```python
# 客製化資料集
import numpy as np
from sklearn.model_selection import train_test_split

# 生成標籤（1 表示正面，0 表示負面），此處隨機設定標籤作為示範
labels = np.random.randint(2, size=len(encoded_data))

# 將數據集分為訓練和測試數據集
x_train, x_test, y_train, y_test = train_test_split(encoded_data, labels, test_size=0.2, random_state=42)

# 檢查分割結果
print(f'訓練集大小：{len(x_train)}，測試集大小：{len(x_test)}')
```

```python
# 客製化資料集
print('訓練總筆數:', len(x_train))
print('測試總筆數:', len(x_test))
```

### 輸入至RNN

- RNN可以處理不同長度的輸入，但還是要注意以下原則
  - 設定輸入文字長度上限
  - 每段文字要一樣長，短的後面補0

```python
from tensorflow.keras.preprocessing import sequence
```

```python
#以tf.keras.preprocessing.sequence.pad_sequences將序列填充到相同的長度。

x_train = sequence.pad_sequences(x_train, maxlen=150)
x_test = sequence.pad_sequences(x_test, maxlen=150)
```

```python
x_train.shape
```

### 建立RNN

這裡我們選用 LSTM，決定神經網路架構
- 先將 10000 維的文字壓到 N 維
- 然後用 K 個 LSTM 神經元做隱藏層
- 最後一個 output, 直接用 sigmoid 送出


建構我們的神經網路
- 文字我們用 1-hot 表示是很標準的方式,不過指定要 1 萬個字, 所以每個字是用 1 萬維的向量表示! 這一來很浪費記憶空間, 二來字和字間基本上是沒有關係的。我們可以用某種「合理」的方式, 把字壓到比較小的維度, 這些向量又代表某些意思 (比如說兩個字代表的向量角度小表相關程度大) 等等。

- 這聽來很複雜的事叫 "word embedding", 而事實上 Keras 會幫我們做。我們只需告訴 Keras 原來最大的數字是多少 (10000), 還有我們打算壓到幾維 (N)。

```python
N = 3 # 文字要壓到 N 維
K = 4 # LSTM 有 K 個神經元
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding
from tensorflow.keras.layers import LSTM

model = Sequential()
model.add(Embedding(10000, N))
model.add(LSTM(K))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

```python
model.summary()
```

### 訓練模型

我們用的 embedding 中, 會被 batch_size 影響輸入。輸入的 shape 會是

(batch_size, 每筆上限)
也就是 (32,100) 輸出是 (32,100,128), 其中 128 是我們決定要壓成幾維的向量。

```python
model.fit(x_train, y_train,
         batch_size=32,
         epochs=5)
```

### 檢視結果

```python
score = model.evaluate(x_test, y_test)
```

```python
print(f'測試資料的 loss = {score[0]}')
print(f'測試資正確率 = {score[1]}')
```

### 儲存結果

```python
model_json = model.to_json()
open('imdb_model_arch.json','w').write(model_json)
model.save_weights('imdb_model_weights.h5')
```

```python
import spacy

# Load the spaCy model
nlp = spacy.load("zh_core_web_sm")

def extract_keywords_spacy(text):
    # Process the text
    doc = nlp(text.lower())
    
    # Filter out stop words and punctuation
    keywords = [token.text for token in doc if not token.is_stop and not token.is_punct]
    
    return keywords

text = "Natural language processing is an exciting field of artificial intelligence."
text = comment
keywords = extract_keywords_spacy(text)
print(keywords)
```

```python
import spacy

nlp = spacy.load("zh_core_web_sm")

def extract_keywords_spacy(text):
    doc = nlp(text.lower())
    keywords = [token.text for token in doc if not token.is_stop and not token.is_punct]
    return keywords

def filter_keywords(keywords, allowed_keywords):
    
    return [word for word in keywords if word in allowed_keywords]

text = comment

allowed_keywords = ['態度', '好', '空間', '狹窄', 'intelligence']
keywords = extract_keywords_spacy(text)
print(keywords)
filtered_keywords = filter_keywords(keywords, allowed_keywords)
print(filtered_keywords)
```
